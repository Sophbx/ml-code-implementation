```python
import numpy as np

# Test matrix
scores = np.array([[85, 90, 78], 
                [92, 88, 86],])

blocked_mask = [0, 0, 1]

# Show the shape of the matrix
print(scores.shape)
```

```python
# transpose of a matrix
print(np.swapaxes(scores, -1, -2))
print(np.transpose(scores))
# The original scores matrix is unchanged
print(scores.shape)
```

```python
# numeric functions
print(np.sqrt(scores))
print(np.sqrt(scores.shape[-1]))

print(np.max(scores, axis=-1, keepdims=True))
print(np.max(scores, axis=0, keepdims=True))
print(np.max(scores))

print(np.exp(scores))

print(np.sum(np.exp(scores), axis=-1, keepdims=True))
print(np.sum(scores, axis=0, keepdims=True))
```

```python
# Turn into numpy array
blocked_mask = np.asarray(blocked_mask, dtype=bool)
print (blocked_mask)
```

```python
# broadcast the mask to the shape of the scores matrix
blocked_mask = np.broadcast_to(blocked_mask, scores.shape)
print(blocked_mask)
```

```python
# Boolean functions
# All true or not, starting with the last axes (col in this case)
print(np.all(blocked_mask, axis=-1))
# starting with the first axes (row in this case)
print(np.all(blocked_mask, axis=0))

# Any true or not, starting with the last axes (col in this case)
print(np.any(blocked_mask, axis=-1))
# starting with the first axes (row in this case)
print(np.any(blocked_mask, axis=0))

# Can group
print(np.any(np.all(blocked_mask, axis=-1)))
print(np.any(np.all(blocked_mask, axis=0)))
```

```python
# Conditional value setting
print(np.where(blocked_mask, -np.inf, scores))
```
