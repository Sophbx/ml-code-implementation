## Q-learning Pseudo Code

for each episode do
    INITIAL STATE <- s
    while s not FINAL STATE do
        # P(exlore - randomly select an action) = ϵ; 
        # P(exploit - Q*(s, a) to select an actioin) = 1−ϵ.
        select ACTION a，use \epsilon-greedy policy
        select ACTION a，obs: r & NEXT STATE s'
        update Q：
            Q(s, a) <- Q(s, a) + \alpha [ r + \gamma max_{a'} Q(s', a') - Q(s, a) ]
        s = s'
    end while
end for
